## Task 1: The Analytical Baseline (Pohlhausen)
Using the Blasius solution for a flat plate boundary layer, solve for the non-dimensional temperature profile $\theta(\eta)$.

1. Numerical Integration: Write a script (Python or MATLAB) to solve the Blasius equation $f''' + \frac{1}{2}ff'' = 0$ with $f(0)=0, f'(0)=0, f'(\infty)=1$.
2. Thermal Solution: Using your result for $f(\eta)$, compute the temperature profile for $Pr = 0.7$ (Air):
   $$\theta(\eta) = \frac{\int_0^\eta \exp\left(-\frac{Pr}{2} \int_0^\eta f \, d\eta\right) d\eta}{\int_0^\infty \exp\left(-\frac{Pr}{2} \int_0^\eta f \, d\eta\right) d\eta}$$
3. Output: Plot $\theta$ vs. $\eta$. Determine the value of $\theta'(0)$, which directly relates to the local Nusselt number.

Note: If you see issues, you should set $\nu = \alpha$. In this case, the thermal boundary layer ($\theta(\eta)$) and viscous boundary layer ($f'(\eta)$) should match.

## Task 4: Darcy-Forchheimer
We cannot model 40 detailed heat sinks in a 1,000-rack data center simulation. We must treat the rack as a porous medium.

1. **Pressure Drop Extraction:** Use your CFD model to measure the pressure drop $\Delta P$ across the heat sink at three different approach velocities ($1.0, 2.0, 3.0 \, \text{m/s}$).

   $$\frac{\Delta P}{L} = \frac{\mu}{\kappa} v + \beta \rho v^2$$
